# 3 Observation

After exploring the dataset using Pandas, the following observations were made:

* The dataset contains **25 rows and 7 columns**.
* The available features are `Student_ID`, `Name`, `Age`, `Course`, `Marks`, `Attendance_Percentage`, and `City`.
* The dataset contains both **numerical and categorical/text features**.
* `Student_ID` and `Age` are integer (`int64`) columns.
* `Marks` and `Attendance_Percentage` are floating-point (`float64`) columns.
* `Name`, `Course`, and `City` are text/object (`object`) columns.
* Missing values are present in the dataset and can be identified using `info()` and `isnull().sum()`.
* The `Marks`, `Attendance_Percentage`, and `City` columns contain missing values.
* The `describe()` function provides useful statistics such as count, mean, standard deviation, minimum, quartiles, and maximum for numerical columns.
* The dataset can be considered a small student-information dataset suitable for practicing basic Pandas data inspection techniques.

Overall, this analysis demonstrates how Pandas can be used to quickly understand the structure, data types, dimensions, descriptive statistics, and missing values of a dataset before performing further analysis.
