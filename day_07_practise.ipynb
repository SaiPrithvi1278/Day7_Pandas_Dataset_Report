## 1. What is Pandas?

**Pandas** is a Python library designed for data manipulation and analysis.

A simple way to think about it:

> **NumPy is excellent for numerical arrays, while Pandas adds labels, tables, and many convenient data-analysis operations.**

Pandas is especially useful when working with data stored in formats such as CSV, Excel, SQL tables, and JSON.

````python
import pandas as pd
````

## 2. Pandas Series

A **Series** is a one-dimensional data structure with:

- values
- an index (labels for the values)

You can think of it as a single column of a table.

### 2.1 Creating a Series from a list

````python
numbers = [1, 2, 3, 4, 5]

s = pd.Series(numbers)
s
````

By default, Pandas creates an integer index starting at `0`.

So the data is conceptually:

| Index | Value |
|---:|---:|
| 0 | 1 |
| 1 | 2 |
| 2 | 3 |
| 3 | 4 |
| 4 | 5 |

### 2.2 Creating a Series with custom index labels

````python
scores = [49, 22, 40, 35, 45]

s = pd.Series(
    scores,
    index=["Python", "SQL", "Pandas", "NumPy", "Statistics"]
)

s
````

Here, the labels are meaningful names instead of `0, 1, 2, ...`.

This is one of the advantages of Pandas: **values can be accessed using meaningful labels.**

### 2.3 Creating a Series from a dictionary

````python
python_scores = {
    "Olii": 49,
    "Maly": 22,
    "Saurab": 40
}

s = pd.Series(python_scores)
s
````

When a dictionary is passed to `pd.Series()`, its **keys become the index** and its **values become the Series values**.

````python
print("Original dictionary:", python_scores)
print("Type of dictionary:", type(python_scores))
print("Type of Series:", type(s))
````

## 3. Pandas DataFrame

A **DataFrame** is a two-dimensional labelled table.

It contains:

- **Rows** — individual records.
- **Columns** — features/attributes.
- **Index** — labels used to identify rows.

A DataFrame is similar to a spreadsheet or a database table.

### 3.1 Creating a DataFrame from a dictionary

````python
cars = {
    "name": ["Audi", "BMW", "MG"],
    "rating": [9, 8.5, 6]
}

df = pd.DataFrame(cars)
df
````

Each dictionary key becomes a column.

For example:

- `name` → one column
- `rating` → another column

The lists provide the values for those columns.

### 3.2 Creating a DataFrame from a NumPy array

````python
import numpy as np

array = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

df = pd.DataFrame(array)
df
````

A NumPy array does not contain column names by default, so Pandas creates the default column labels `0` and `1`.

### 3.3 Converting a DataFrame to a NumPy array

````python
numpy_array = df.to_numpy()
numpy_array
````

`to_numpy()` converts the DataFrame's values into a NumPy array.

This is useful when you need to pass the numerical data to code or libraries that expect NumPy arrays.

## 4. Basic Series Operations

Pandas Series support many normal arithmetic operations.

The important idea is that Pandas performs operations while respecting the Series index.

````python
s1 = pd.Series(
    [10, 20, 30],
    index=["a", "b", "c"]
)

print("Original Series:")
print(s1)

print("\nSeries multiplied by 2:")
print(s1 * 2)
````

### 4.1 Series operations align by index

````python
s2 = pd.Series(
    [1, 3, 4],
    index=["a", "c", "d"]
)

print("s1:")
print(s1)

print("\ns2:")
print(s2)

print("\ns1 + s2:")
print(s1 + s2)
````

Notice that Pandas matches values using their **index labels**:

- `a` exists in both Series → values are added.
- `c` exists in both Series → values are added.
- `b` exists only in `s1` → result is `NaN`.
- `d` exists only in `s2` → result is `NaN`.

This behaviour is called **index alignment**.

Once a dataset has been loaded, we need to understand its structure.

These commands are among the most important Pandas tools for a beginner:

|---|---|

The next sections explain each one.

### 9.1 `shape` — number of rows and columns

````python
# df.shape
````

The result is a tuple:


For example, `(100, 7)` means:

- 100 rows
- 7 columns

### 9.2 `columns` — column names

````python
# df.columns
````

This shows all column labels in the DataFrame.

### 9.3 `index` — row index information

````python
# df.index
````

For a normal DataFrame, you may see something similar to:

```text

```

This means the index starts at `0`, ends before `100`, and increases by `1`.

### 9.4 `dtypes` — data types

````python
# df.dtypes
````

Common Pandas data types include:

- `int64` → integer values
- `float64` → decimal values

````python
# df.info()
````

`info()` is extremely useful because it combines several pieces of information:

- number of rows
- column names

- data types

### 9.6 `describe()` — descriptive statistics

````python
# df.describe()
````

For numerical columns, `describe()` normally provides:

- `count`
- `mean`
- `std`
- `min`
- `25%`
- `50%` (median)
- `75%`
- `max`

For example, if a `Marks` column has a mean of `80`, the average mark is 80.

`describe()` is a quick way to understand the distribution and range of numerical data.

### Describing categorical columns

You can also ask `describe()` to include non-numerical columns:

````python
# df.describe(include="all")
````

For categorical/text columns, useful statistics can include:

- `count` → number of non-missing values
- `unique` → number of distinct values
- `top` → most frequent value
- `freq` → frequency of the most frequent value

## 10. Checking Missing Values

`info()` can show whether the number of non-null values is smaller than the total number of rows.

You can also check missing values directly:

````python
# df.isna().sum()
````

`isna()` identifies missing values, and `sum()` counts them column by column.

For example:

```text
Marks       1
City        0
Age         0
```

would mean that one value is missing from `Marks`, while `City` and `Age` have no missing values.

## 11. Common Pitfalls When Loading Data

Loading a file successfully does **not** automatically mean that it was loaded correctly.

Always check the data after loading it.

### Common problems

1. **Wrong separator**
   - The file may use `,`, `;`, or tabs.

2. **Missing values**
   - Some cells may be empty or contain special missing-value markers.

3. **Incorrect header handling**
   - The first row may or may not contain column names.

4. **Wrong index**
   - A column may accidentally be used as the index, or the index may have been written to the CSV.

5. **Unexpected data types**
   - A column that should contain numbers may be read as text because of inconsistent values.

### A good first inspection sequence

After loading a new CSV, a useful first check is:

```python
df.head()
df.tail()
df.sample()
df.shape
df.columns
df.index
df.dtypes
df.info()
df.describe()
df.isna().sum()
```
